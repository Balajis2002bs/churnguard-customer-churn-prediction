```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("data\\raw\\Telco-Customer-Churn.csv")
df.head()
```

### Removing Customer ID columns

```python
df=df.drop(columns=['customerID'])
print(df.columns)
```

### Removing Duplicate Rows

```python
df=df.drop_duplicates()
df.duplicated().sum()
```

### Striping the whitespaces from  gender and PaymentMethod

```python
cat_cols=df.select_dtypes('object').columns
for col in cat_cols:
    df[col]=df[col].str.strip().str.title()
```

### Maping the Contract Values into Three Categories such as Month-to-month, One year, Two year

```python
df['Contract']=df['Contract'].str.lower()
df['Contract']=df['Contract'].map({
  
    'month to month': 'Month-to-month',
    'month-to-month': 'Month-to-month',
    'monthly':'Month-to-month',
    '1 year': 'One year',
    'one year': 'One year',
    '2 year': 'Two year',
    'two year': 'Two year',
    
    
})


print(f"After Cleaning, Value counts for 'Contract': {df['Contract'].value_counts().to_dict()}")
```

### Maping InternetService into three Categories such as DSL, Fiber optic, No

```python
print(df['InternetService'].value_counts().to_dict())
df['InternetService']=df['InternetService'].str.strip().str.lower()
print(df['InternetService'].value_counts().to_dict())
print(df['InternetService'].unique())
print(df['InternetService'].isna().sum())




print(df[df['InternetService'].isna()])
```

```python
print('Before Mapping InternetService: ')
print(df['InternetService'].value_counts())
print()
df['InternetService']=df['InternetService'].str.strip().str.lower()
df['InternetService']=df['InternetService'].map(
    {
      'dsl': 'DSL',
      'fiber optic': "Fiber optic",
      'fiberoptic':'Fiber optic',
      'fibre optic':'Fiber optic',
      'NaN' : 'No',
      'no' : 'No',
      
    }
)
df['InternetService']=df['InternetService'].fillna('No')
print()

print('After Mapping InternetService: ')
print(df['InternetService'].value_counts())


```

### Converting Total charges to numeric dtype

```python
df['TotalCharges']=pd.to_numeric(df['TotalCharges'],errors='coerce')
print(df['TotalCharges'].dtype)
print()
print(df['TotalCharges'].isna().sum())
```

### Removing rows from Tenure Columns which have zero or negative value

```python
df=df.drop(df[df['tenure'] <=0].index )
```

###  Removing rows from MonthlyCharges Columns which have value less than 10 and greater than 200

```python
df=df.drop(df[(df['TotalCharges'] <= 10 )& (df['TotalCharges'] >= 20)].index )
```

### Filling missing values in Monthly Charges, Tenure and Total Charges Columns

```python
#Filling mean value for Monthly Charges and Total Charges
for col  in ['MonthlyCharges','TotalCharges']:
    df[col]=df[col].fillna(df[col].mean())

#Filling median value for Tenure

df['tenure']=df['tenure'].fillna(df['tenure'].median().__round__(2))



```

### Shape of the Dataset After Cleaning and Row Counts to Confirm the cleaning is  Complete

```python
print('After Cleaning: ')
print()
print(f'In Cleaned Dataset, Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}')
print()


print(f'In Cleaned Dataset, Total Missing Values: {df.isna().sum()}')

```

### Storing the cleaned Dataset into data/processed folder

```python
df.to_csv("data\\processed\\1cleaned_churnguard_data_cleaned.csv",index=False)
```
